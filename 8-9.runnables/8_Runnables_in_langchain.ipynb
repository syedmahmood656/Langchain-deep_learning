In this notebook we will be creating 3 langchain components **(1.model(LLM), 2.prompt_template, 3.chain )** from scratch to  deeply understand how these components works, later we will code runnables which seemlessly connect these different compontes and generates efficent responses

```
import random
```

here we made a class that acts as a "LLM" but a dummy one, which gives responses randomly from a variable (response_list)

when we call predict function from dummy_llm one of four pre-written responses generates as output randomly

```
class Dummy_llm:
  def __init__(self):
    print("llm created")

  def predict(self,prompt):

    print("prompt: ",prompt)

    response_list = [
        "AI stands for artificial intellegence",
        "AI is a domain in CS",
        "AI is now the most trending tiopic in IT",
        "Every bussniss now wants AI in their applications"
    ]

    return {"response": (random.choice(response_list))}
```

below dummy prompt template acts like prompt template in langchain, it helps create a template and a template variable, this is basically how langchains prompt template works and coded in library

```
class Dummy_promot_template:
  def __init__(self,template,template_variable):
    self.template = template
    self.template_variable = template_variable

  def format(self,input_dict):
    return self.template.format(**input_dict)
```

below we are making a prompt template using above dummy prompt template class

```
template = Dummy_promot_template(
    template = "give a fact about {topic}",
    template_variable = ["topic"]
)
```

we are assigning the value to template variable

```
prompt = template.format({'topic':'AI'})
```

```
llm = Dummy_llm()
```

Here our dummy prompt template assigne the variable and our dummy_llm generates the resonse

```
llm.predict(prompt)
```

```
class Dummy_llm_chain:
  def __init__(self,llm,prompt):
    self.llm = llm
    self.prompt = prompt

  def run(self,input_dict):
    final_prompt = self.prompt.format(input_dict)
    result =  self.llm.predict(final_prompt)

    return result['response']
```

```
template = Dummy_promot_template(
    template = "give a fact about {topic}",
    template_variable = ["topic"]
)
```

```
llm = Dummy_llm()
```

```
chain = Dummy_llm_chain(llm, template)
```

Here we built chain componet that works simlar to chains in langchain

```
chain.run({'topic':'AI'})
```

here we will be implementing the runnables concept of langchain where we will
1. standardizing components
2. should contain same methods "invoke" -> (used concept of abstraation from OOPs)

we will ensure all the components must have same methods using abstract class named "runnables" where all the components will inherits from this class and in result we will get same methods

```
from abc import ABC, abstractmethod
```

```
class Runnable(ABC):

  @abstractmethod
  def invoke(input_data):
    pass
```

```
class Dummy_llm(Runnable):
  def __init__(self):
    print("llm created")

  def predict(self,prompt):

    print("prompt: ",prompt)

    response_list = [
        "AI stands for artificial intellegence",
        "AI is a domain in CS",
        "AI is now the most trending tiopic in IT",
        "Every bussniss now wants AI in their applications"
    ]

    return {"response": (random.choice(response_list))}
```

As you can see we will get error if we dont use or mention invoke method in dummyllm class as we have made dummyllm class as child class of Runnable class

```
llm = Dummy_llm()
```

now we are writing "invoke" method in our class by force to ensure same methods in components and enusre connectivity while connecting different components

```
class Dummy_llm_2(Runnable):
  def __init__(self):
    print("llm created")

  def invoke(self,prompt):

     print("prompt: ",prompt)

     response_list = [
         "AI stands for artificial intellegence",
        "AI is a domain in CS",
        "AI is now the most trending tiopic in IT",
        "Every bussniss now wants AI in their applications"
     ]
     return {"response": (random.choice(response_list))}

  def predict(self,prompt):

    print("prompt: ",prompt)

    response_list = [
        "AI stands for artificial intellegence",
        "AI is a domain in CS",
        "AI is now the most trending topic in IT",
        "Every bussniss now wants AI in their applications"
    ]

    return {"response": (random.choice(response_list))}
```

```
class Dummy_promot_template(Runnable):
  def __init__(self,template,template_variable):
    self.template = template
    self.template_variable = template_variable

  def invoke(self,input_dict):
    return self.template.format(**input_dict)

  def format(self,input_dict):
    return self.template.format(**input_dict)
```

now it runs as there is invoke method in dummy_llm_2

```
llm = Dummy_llm_2()
```

now we will make more components of langchain like string parser

```
class Dummy_StrOutputParser(Runnable):

  def __init__(self):
    pass

  def invoke(self, input_data):
    return input_data['response']
```

Now, finally we will code a **runnable connector class** that acts like a chain that connects all the compones seemlessly.!!

```
class Runnable_connerctor(Runnable):
  def __init__(self,runnable_list):
    self.runnable_list = runnable_list

  def invoke(self,input_data):
    for Runnable in self.runnable_list:
      input_data = Runnable.invoke(input_data)

    return input_data
```

```
template = Dummy_promot_template(
    template = "give a fact about {topic}",
    template_variable = ["topic"]
)
```

```
parser = Dummy_StrOutputParser()
```

```
chain = Runnable_connerctor([template, llm, parser])
```

```
chain.invoke({'topic':'AI'})
```

Now we have finally implemented the concept of runnables/conntectors through standardzing methods in the components of langchain (dummy), thats how the developers of langchain solved the problem of chains and came up with concept of runnables
