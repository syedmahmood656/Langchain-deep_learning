```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

### creating tool

```python
# creating a tool by simply writing "@tool" on the function as decorator which tells our llm that this is a tool
@tool
def multiply(a: int, b: int) -> int:
  """Given 2 numbers a and b this tool returns their product"""
  return a * b
```

```python
# using invoke as multiply is a tool and a runnable as we have use tool decorator "@tool"
print(multiply.invoke({'a':3, 'b':4}))
```

```python
print(multiply.description)
print(multiply.args)
print(multiply.schema)
# we can see the discription of the tool, arguments, type and schemas
```

### Tool Binding

```python
!pip install langchain_huggingface
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace # imporing required libraryes
```

```python
# setting up hugging face free chatmodel to use for text generation

import os
# Set the token in environment variables specifically for the hub client
os.environ['HUGGINGFACEHUB_API_TOKEN'] = '=Your API key'

llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    huggingfacehub_api_token=os.environ['HUGGINGFACEHUB_API_TOKEN'],
    temperature=0.2
)

# Some wrappers need the token explicitly passed if environment sync is slow
model = ChatHuggingFace(llm=llm, huggingfacehub_api_token=os.environ['HUGGINGFACEHUB_API_TOKEN'])
```

```python
llm_with_tools =  model.bind_tools([multiply]) # here we have given this specific llm a tool ## tool binding has been done here
```

### tool calling

```python
llm_with_tools.invoke('hi how are you') # this will return a genral rresponse from llm
```

```python
llm_with_tools.invoke('multiply 3 with 40') # this will return with a message that suggests tools and its parameters hense known as tool calling
```

### tool execution

```python
result = llm_with_tools.invoke('multiply 3 with 40')  # store the result
```

```python
multiply.invoke(result.tool_calls[0]) # this will return a tool message and the result of multiplication of two givven numbers
```
